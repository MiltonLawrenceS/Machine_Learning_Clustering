Load the Scaling,Model & Cluster_Dataset:

```python
import pickle
loaded_sc=pickle.load(open('StandardScaler.sav','rb'))
loaded_model=pickle.load(open('finalized_model.sav','rb'))
final_loaded_dataset=pickle.load(open('Deployment_dataset.sav','rb'))
```

Final Prediction:

```python
Salary = int(input("Enter the Annual Salary(k$):"))
Spending = int(input("Enter the Spending Score(1-100):"))
new_customer = [[Salary, Spending]]
new_scaled = loaded_sc.transform(new_customer)
prediction = loaded_model.predict(new_scaled)

print("User Entered Salary:", Salary)
print("User Entered Spending Amount:", Spending)
print("--------------------------------")

if prediction == 0:
    print("This customer will earn a moderate salary and maintain a moderate level of spending.")
elif prediction == 1:
    print("This customer will earn a moderate to high salary and maintain a high level of spending.")
elif prediction == 2:
    print("This customer will earn a moderate to high salary and maintain a low level of spending, indicating a saving tendency.")
elif prediction == 3:
    print("This customer will earn a low salary but maintain a relatively high level of spending.")
elif prediction == 4:
    print("This customer will earn a low salary and maintain a low level of spending.")
elif prediction == 5:
    print("This customer will earn the highest salary and maintain a high spending tendency.")
else:
    print("This customer will earn a high salary but maintain a very low level of spending, indicating a high saving tendency.")
```
